#Шаг 1  
Найдите первую и вторую производные функции y в заданной точке x0
для своего варианта с помощью scipy.  
Вариант 2: y=ln(sqrt(x)), x_0 = 4, a = 3, b = 8

```python
import numpy as np
from scipy.misc import derivative

def f(x): return np.log(np.sin(np.sqrt(x)))
print(derivative(f, 4, n=1))
print(derivative(f, 4, n=2))
```

#Шаг 2  
Используя функцию из таблицы выше для вашего варианта и пакет
sympy выведите на экран символьное представление производной.

```python
import sympy as sp
from IPython.display import display

x = sp.symbols("x")
func = sp.log(sp.sin(sp.sqrt(x)))
der = sp.diff(func, x)
display(der)
```

#Шаг 5  
Решите задачу нелинейной оптимизации для вашего варианта.
Выведите оптимальное значение и решение.  
![Задача оптимизации, второй вариант](image.png)

```python
from scipy.optimize import minimize, Bounds

def L(x): return (x[0] - 3) ** 2 + (x[1] - 1) ** 2
def g1(x): return -2 * x[0] + x[1] - 2
def g2(x): return 3 * x[1] - 10

bounds = Bounds([0, 0], [np.inf, np.inf])
constraints=[{"type": "ineq", "fun": g1}, {"type": "ineq", "fun": g2}]

print(minimize(L, [0,0], bounds=bounds, constraints=constraints))
```
