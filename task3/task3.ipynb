#Шаг 1  
Взять свою матрицу из 1.10

```python
import numpy as np 
import scipy as sc
matrix = np.array([[2, -5, 1, 0], [1, -1, -13, 0], [3, -2, -2, -4], [4, 0, 2.7, -1.3]])
```

#Шаг 2  
Получить LU-разложение (LUP-разложение) (факторизацию матрицы)


```python
P, L, U = sc.linalg.lu(matrix)
print(P @ L @ U == matrix)
```

#Шаг 3  
Найти определитель (произведение элементов на главной диагонали L и
U, определителя обратной матрицы к P)

```python
det = np.linalg.det(np.linalg.inv(P)) * np.prod(np.diag(L)) * np.prod(np.diag(U))
print(det, np.linalg.det(matrix))
```

#Шаг 4  
Сгенерировать две выборки (вектор целых чисел из 100 элементов) в
одинаковых интервалах с разными распределениями (равномерное и
нормальное)

```python
vec_1 = np.random.uniform(0, 31, (100)).astype(int)
vec_2 = np.random.normal(15, 10, (100)).astype(int)
vec_2 = np.clip(vec_2, 0, 30)
print(vec_1[0:5], vec_2[0:5])

vec_1 = sc.stats.uniform.rvs(0, 31, 100).astype(int)
vec_2 = sc.stats.norm.rvs(15, 10, 100).astype(int)
vec_2 = np.clip(vec_2, 0, 30)
print(vec_1[0:5], vec_2[0:5])
```

#Шаг 5  
Вычислить для каждой из выборок:  
a. Среднее;  
b. Моду;  
c. Медиану;  
d. Минимум;  
e. Максимум;  
f. Стандартное отклонение.

```python
print(f"Равномерное | Нормальное")
print(f"Среднее значение: {np.mean(vec_1)} | {np.mean(vec_2)}")
print(f"Мода: {sc.stats.mode(vec_1)[0]} | {sc.stats.mode(vec_2)[0]}")
print(f"Медиана: {np.median(vec_1)} | {np.median(vec_2)}")
print(f"Минимум: {np.min(vec_1)} | {np.min(vec_2)}")
print(f"Максимум: {np.max(vec_1)} | {np.max(vec_2)}")
print(f"Стандартное отклонение: {np.std(vec_1)} | {np.std(vec_2)}")
```
