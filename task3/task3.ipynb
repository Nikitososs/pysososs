#Шаг 1  
Взять свою матрицу из 1.10

```python

def mprint(matrix):
    print('\n'.join('   '.join(str(x) for x in row) for row in matrix))
```

```python
import numpy as np 
import scipy as sc
matrix = np.array([[2, -5, 1, 0], [1, -1, -13, 0], [3, -2, -2, -4], [4, 0, 2.7, -1.3]])
mprint(matrix)
```

#Шаг 2  
Получить LU-разложение (LUP-разложение) (факторизацию матрицы)


```python
L, U = sc.linalg.lu(matrix, permute_l=True)
np.allclose(L @ U, matrix)
```

```python
P, L, U = sc.linalg.lu(matrix)
np.allclose(P @ L @ U, matrix)
```

#Шаг 3  
Найти определитель (произведение элементов на главной диагонали L и
U, определителя обратной матрицы к P)

```python
det = np.linalg.det(np.linalg.inv(P)) * np.prod(np.diag(L)) * np.prod(np.diag(U))
np.allclose(det, np.linalg.det(matrix))
```

С помощью метода хи-квадрат (scipy.stats.chisquare) вычислить значение
p-value для нулевой гипотезы: “Распределение выборки не
равномерное”.

chisquare проверяет соответствие наблюдаемых данных ожидаемым, предположение:  
задать ожидаемой равномерную выборку и проверить обратное, нулевую гипотезу "выборка равномерна".

# Для изначально заданного нормального распределения

```python
f_obs = sc.ndimage.histogram(uniform, 0, 30, 30)
f_exp = [np.sum(f_obs) / 30] * 30
statistic, pvalue = sc.stats.chisquare(f_obs, f_exp)
print(pvalue)
```

Т.к. p-значение = 0.8..., что больше 0,05,  это значит нулевая гипотеза о том, что выборка равномерна верна
Гипотеза совпадает с тем, что изначальная выборка действительно является равномерным распределеним

chisquare проверяет соответствие наблюдаемых данных ожидаемым, предположение:  
задать ожидаемой равномерную выборку и проверить обратное, нулевую гипотезу "выборка равномерна".

Выборка будет представленна баллами абитуриентов, поступивших в ОмГТУ на направление 02.03.02 в этом году.

Выделим три категории по принципу:  
Разницу между баллами абитуриента с наибольшим чилом балов и абитуриента с наименьшим числом баллов, поделим на три, и опираясь на это выделим следующие категории:  
1. Абитуриентов с баллами ниже средних  
1. Абитуриентами со средними баллами  
1. Абитуриентов с высокими баллами  

И проверим эту выборку на равномерность.

```python
def prepare():
    entrants = [253, 254, 253, 245, 261, 262, 239, 252, 257, 274, 257, 252, 259, 254, 244, 240, 246, 239, 263, 236, 235, 240, 240, 256, 265, 250, 234, 237, 250, 257, 248, 240, 242, 253, 235, 242, 259, 267, 244, 253, 238, 237, 248, 242, 254, 264]
    
    mins, maxs= np.min(entrants), np.max(entrants)
    cat_step = (maxs - mins) / 3

    cat1_lb, cat1_ub = mins, mins + cat_step
    cat2_lb, cat2_ub = cat1_ub, cat1_ub + cat_step
    cat3_lb, cat3_ub = cat2_ub, maxs

    cat1_cnt = len([abit for abit in entrants if cat1_lb <= abit < cat1_ub])
    cat2_cnt = len([abit for abit in entrants if cat2_lb <= abit < cat2_ub])
    cat3_cnt = len([abit for abit in entrants if cat3_lb <= abit <= cat3_ub])

    return [cat1_cnt, cat2_cnt, cat3_cnt], [len(entrants) / 3] * 3
```

```python
f_obs, f_exp = prepare()
statistic, pvalue = sc.stats.chisquare(f_obs, f_exp)
print(pvalue)
```

Т.к. p-значение = 0.03..., что меньше 0,05, значит нулевая гипотеза о том, что выборка равномерна неверна, следовательно верна обратная к ней: выборка неравномерна
