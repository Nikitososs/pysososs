#Шаг 1  
Используйте функцию и отрезок ([a, b]) из вашего варианта задания 2
(которые вы использовали для интегрирования). Разбейте отрезок на
столько частей, сколько считаете нужным. Вычислите значение
функции для всех точек получившегося массива.  
Отрезок: a = 1, b = 6  
Функция: y=ln(sin(sqrt(x)))

```python
import numpy as np
import scipy as sc

dots = np.arange(1, 6, 0.25)
def f(x): return np.log(np.sin(np.sqrt(x)))
result = [f(x) for x in dots]
print(*result)
```

#Шаг 2  
Используя множество точек и множество значений функции в этих
точках, постройте простой график функции (метод plot()). Задайте
графику название. Подпишите оси. Установите желаемые размеры
шрифта для названия и подписей. Добавьте на график легенду.

```python
import matplotlib.pyplot as plt

plt.plot(dots, result, label="y=ln(sin(sqrt(x)))")
plt.title("График функции").set_fontsize(16)
plt.xlabel('Ось X').set_fontsize(12)
plt.ylabel('Ось Y').set_fontsize(12)
plt.legend(loc="lower left")
plt.show()
```

#Шаг 6  
Используя знания о разбиении отрезка на множество точек (из пункта
1), а также функцию с несколькими переменными из задания 2 (которая
использовалась для решения задачи оптимизации), постройте
трехмерный график функции. Установите цвет линии на графике.
Важно: если в вашем варианте к заданию 2 функция состояла более чем
из двух переменных - используйте только первые две переменные.

```python
X = np.arange(1, 6, 0.25)
Y = np.arange(1, 6, 0.25)
X, Y = np.meshgrid(X, Y)

def L(x1, x2): return (x1 - 3) ** 2 + (x2 - 1) ** 2
Z = L(X, Y)

fig = plt.figure(figsize=(15, 15))
ax = fig.add_subplot(projection='3d')
ax.plot_surface(X, Y, Z, color="r")
ax.set_xlabel('Ось X').set_fontsize(12)
ax.set_ylabel('Ось Y').set_fontsize(12)
ax.set_zlabel('Ось Z').set_fontsize(12)
plt.title("Целевая функция")
plt.show()
```
