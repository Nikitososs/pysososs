Используйте функцию и отрезок ([a, b]) из вашего варианта задания 2

функции для всех точек получившегося массива.  
Отрезок: a = 1, b = 6  
Функция: y=ln(sin(sqrt(x)))

```python
import numpy as np
import scipy as sc

dots = np.arange(1, 6, 0.01)
print(len(dots))
def f(x): return np.log(np.sin(np.sqrt(x)))
result = [f(x) for x in dots]
print(*result)
```

#Шаг 7  
Используя подграфики (subplots) постройте сетку 2x2 из 4 графиков.
Установите размер сетки. Добавьте заголовок сетки, укажите размер
шрифта. В левом верхнем углу сетки разместите график из пункта 2
текущего задания, в правом верхнем - из пункта 3. В левом нижнем -
круговую диаграмму из пункта 5. В правом нижнем - трехмерный
график из пункта 6. Для любого из графиков установите цвет фона.

```python
from typing import Literal
from matplotlib.figure import Figure
from matplotlib.axes import Axes

class TextOptions:
    text: str | None = None
    font_size: int = 16

    def __init__(self, text: str |None = None, font_size: int = 16):
        self.text = text
        self.font_size = font_size    

class GridOptions:
    enabled: bool = False
    color: tuple | None = None
    alpha: float | None = None

    def __init__(self, enabled: bool = False, color: tuple | None = None, alpha: float | None = None):
        self.enabled = enabled
        self.color = color
        self.alpha = alpha

def make_plt(
    plot_type: Literal["plot", "scatter", "pie", "surface"],
    display_data: tuple,
    ax: Axes | None = None, 
    title: TextOptions | None = None,
    x_label: TextOptions | None = None,
    y_label: TextOptions | None = None,
    figure: Figure | None = None,
    label: str | None = None,
    legend_loc: str | None = None,
    autopct: callable = lambda x : x,
    color: tuple | None = None,
    marker: str | None = None,
    grid: GridOptions | None = None,
    face_color: tuple | None = None,
    labels: list | None = None,
    colors: list | None = None,
    cmap: str | None = None,
):
    
    if plot_type == "plot":
        if ax is not None:
            ax.plot(*display_data, label=label)
            title and ax.set_title(title.text).set_fontsize(title.font_size)
            x_label and ax.set_xlabel(x_label.text).set_fontsize(y_label.font_size)
            y_label and ax.set_ylabel(y_label.text).set_fontsize(y_label.font_size)
            legend_loc and ax.legend(loc=legend_loc)
    
    if plot_type == "scatter":
        if ax is not None:
            ax.scatter(*display_data, label=label, marker=marker, color=color)
            title and ax.set_title(title.text).set_fontsize(title.font_size)
            x_label and ax.set_xlabel(x_label.text).set_fontsize(y_label.font_size)
            y_label and ax.set_ylabel(y_label.text).set_fontsize(y_label.font_size)
            legend_loc and ax.legend(loc=legend_loc)
            grid and ax.grid(grid.enabled, color=grid.color, alpha=grid.alpha)
            face_color and ax.set_facecolor(color=face_color)

    if plot_type == "pie":
        if ax is not None:
            ax.pie(display_data, labels=labels, autopct=autopct, colors=colors)
            ax.set_title(title.text)

    if plot_type == "surface":
        if figure is not None:
            ax and figure.delaxes(ax)
            ax = figure.add_subplot(2, 2, 4, projection='3d')
            ax.plot_surface(*display_data, color=color, cmap=cmap)
            ax.set_title(title.text)
```

```python
def step_7():
    fig, ((ax1, ax2), (ax3, ax4)) = plt.subplots(2, 2, figsize=(15, 15))
    fig.suptitle("Сетка 2 на 2", fontsize=20)

    make_plt(
        plot_type="plot", 
        ax=ax1, 
        display_data=(dots, result), 
        title=TextOptions("График функции", 16), 
        x_label=TextOptions('Ось X', 12), 
        y_label=TextOptions('Ось Y', 12),  
        label="y=ln(sin(sqrt(x)))", 
        legend_loc="lower left"
    )
    make_plt(
        plot_type="scatter", 
        ax=ax2, 
        display_data=(dots, result), 
        title=TextOptions("График функции", 16), 
        x_label=TextOptions('Ось X', 12),
        y_label=TextOptions('Ось Y', 12), 
        marker="*", 
        color=(0.5, 0.4, 0.3), 
        grid=GridOptions(True, color=(0.9, 0.4, 0.3), alpha=0.5), 
        face_color=(0.4, 0.8, 0.2)
    )
    make_plt(
        plot_type="pie", 
        ax=ax3, 
        display_data=data[1], 
        title=TextOptions("Круговая диаграмма", 16), 
        x_label=TextOptions('Ось X', 12),
        y_label=TextOptions('Ось Y', 12), 
        labels = ["Единицы", "Двойки", "Тройки", "Четвёрки"], 
        colors = ['#ff6b6b', '#4ecdc4', '#96ceb4', '#feca57'], 
        autopct = lambda pct: f'{int(round(pct*sum(data[1])/100.0))}'
    )
    make_plt(
        plot_type="surface", 
        ax=ax4, 
        figure=fig,
        display_data=(X,Y,Z), 
        title=TextOptions("Целевая функция"), 
        color="b",
        cmap='viridis'
    )
    plt.show()

step_7()
```
