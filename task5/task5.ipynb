#Шаг 11  
Добавьте колонку “BirthYear” с годом рождения каждого пассажира.

```python
def calc_age(age):
    if pd.isna(age):
        return np.nan
    return 1912 - age
```

```python
titanic_df.insert(11, "BirthDate", [calc_age(age) for age in titanic_df.age.array])
print(titanic_df.sort_values(by="age", ascending=True).head(1))
titanic_df
```

#Шаг 16  
Отобразите сколько женщин и мужчин выжило/не выжило.

```python
female_survived = len(titanic_df[(titanic_df.sex == 'f') & (titanic_df.survived == 1)])
female_dead = len(titanic_df[(titanic_df.sex == 'f') & (titanic_df.survived == 0)])
male_survived = len(titanic_df[(titanic_df.sex == 'm') & (titanic_df.survived == 1)])
male_dead = len(titanic_df[(titanic_df.sex == 'm') & (titanic_df.survived == 0)])

print(f"Выживщие женщины: {female_survived}, невыжившие женщины: {female_dead}")
print(f"Выживщие мужчины: {male_survived}, невыжившие мужчины: {male_dead}")

survival_by_sex = titanic_df.groupby(['sex', 'survived']).size().unstack(fill_value=0)
print(survival_by_sex)
```
