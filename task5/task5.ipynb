#Шаг 9  
Найдите билеты по которым плыло 6 и более человек, отобразите список людей
(groupby, count).

```python
for ticket, dframe in titanic_df.groupby(by="ticket"):
    if dframe.name.count() >= 6:
        print(f'{dframe.name}\n-----')
```
